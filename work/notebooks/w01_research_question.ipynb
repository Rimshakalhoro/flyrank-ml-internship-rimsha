### Research question

Which pages should be reviewed first for a possible content refresh or other content action, based on observable search-performance signals?

### Decision

The decision is which pages deserve limited content-team review first.

### Unit of analysis

The unit of analysis is one content page.

### Output

The intended output is a ranked list of pages for human review, with a score and simple reason codes explaining why each page was prioritized.

### Action

A content or SEO reviewer could review the highest-priority pages first and decide whether to refresh, expand, protect, prune, or monitor them.

A false positive could cause the team to spend time changing a page that did not need attention. A false negative could cause a genuinely weak or declining page to be missed while the team spends its limited time elsewhere. Therefore, the final system should support prioritization rather than automatically deciding what action must be taken.
